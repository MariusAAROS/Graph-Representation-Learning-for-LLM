# KGQA Datasets Explorer

This notebook explores various Knowledge Graph Question Answering (KGQA) datasets.

```python
import datasets
import json
```

## 1. ComplexQuestions Dataset

The ComplexQuestions dataset contains complex questions that require multi-hop reasoning over knowledge graphs. Questions may require combining multiple constraints and relationships.

```python
from data.KGQA.complex_questions.complex_questions import ComplexQuestions
```

```python
builder = ComplexQuestions(config_name="complex_questions")
builder.download_and_prepare()
ds = builder.as_dataset()
```

```python
ds["train"][0]
```

## 2. LC-QuAD 2.0 Dataset

LC-QuAD 2.0 is a large-scale complex question answering dataset with 30,000 QA pairs over Wikidata and DBpedia. Each question has a corresponding SPARQL query, paraphrases, and template information.

```python
from data.KGQA.lcquad_v2.lcquad_v2 import LCQuAD2

builder = LCQuAD2(config_name="lcquad2-wikidata")
builder.download_and_prepare()
ds = builder.as_dataset()
```

```python
ds["train"][3]
```
