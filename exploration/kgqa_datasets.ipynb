```python
# Method 1: Direct data loading without loading scripts (Recommended)
# Since loading scripts are deprecated, load data directly from URLs
import urllib.request
import ast
from pathlib import Path

# URLs from the dataset builder
data_urls = {
    "train": "https://raw.githubusercontent.com/JunweiBao/MulCQA/ComplexQuestions/compQ.train.release",
    "test": "https://raw.githubusercontent.com/JunweiBao/MulCQA/ComplexQuestions/compQ.test.release"
}

# Create data directory
data_dir = Path("./data/KGQA/complex_questions")
data_dir.mkdir(parents=True, exist_ok=True)

def load_complex_questions_dataset(data_urls):
    """Load ComplexQuestions dataset directly from URLs"""
    dataset = {"train": [], "test": []}
    
    for split, url in data_urls.items():
        output_path = data_dir / f"compQ.{split}.release"
        
        # Download if not exists
        if not output_path.exists():
            print(f"Downloading {split} data...")
            urllib.request.urlretrieve(url, output_path)
        
        # Parse the file
        print(f"Loading {split} split...")
        with open(output_path, 'r', encoding='utf-8') as f:
            for line in f:
                line = line.strip()
                if not line:
                    continue
                parts = line.split('\t')
                if len(parts) >= 2:
                    question = parts[0]
                    answers = ast.literal_eval(parts[1]) if len(parts) > 1 else []
                    dataset[split].append({
                        "question": question,
                        "answers": answers
                    })
        
        print(f"✓ Loaded {len(dataset[split])} examples from {split} split")
    
    return dataset

# Load the dataset
try:
    dataset = load_complex_questions_dataset(data_urls)
    print("\n✓ Dataset loaded successfully!")
except Exception as e:
    print(f"✗ Error loading dataset: {e}")
    dataset = None
```
